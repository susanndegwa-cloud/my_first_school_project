## Data Cleaning and Preprocessing 
The dataset is inspected before modelling. The checks below report missing values, duplicate rows, column data types, and potential outliers. Since the supplied dataset has no missing values or duplicate rows, no rows are removed or values imputed. Potential outliers are flagged for review rather than automatically deleted.

```python
dataset = boston

# 1. Check the dataset dimensions and column data types
print('Dataset shape:', dataset.shape)
print('\nColumn data types:')
print(dataset.dtypes)

# 2. Check for missing values
print('\nMissing values per column:')
print(dataset.isnull().sum())
print('Total missing values:', dataset.isnull().sum().sum())

# 3. Check for duplicate rows
print('\nDuplicate rows:', dataset.duplicated().sum())
```

### Outlier inspection
The IQR method is used to flag unusually low or high values in continuous numeric columns. Binary (`CHAS`) and discrete (`RAD`) variables are excluded from this particular check because the IQR rule is less informative for these coded categories. Flagged observations are retained for now; they should only be removed or adjusted if there is evidence that they are errors.

```python
# 4. Identify potential outliers using the IQR rule
continuous_cols = [col for col in dataset.select_dtypes(include=np.number).columns
                   if col not in ['CHAS', 'RAD']]
Q1 = dataset[continuous_cols].quantile(0.25)
Q3 = dataset[continuous_cols].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outlier_counts = ((dataset[continuous_cols] < lower_bound) |
                  (dataset[continuous_cols] > upper_bound)).sum()
print('Potential outliers per continuous column:')
print(outlier_counts)
```

### Preprocessing decisions
- No missing-value treatment is needed because the inspection found no missing values.
- No duplicate removal is needed because no duplicate rows were found.
- Potential outliers are reported and retained so that valid extreme housing observations are not discarded without justification.
- The dataset columns are already numeric, so no categorical text encoding or data-type conversion is required for this dataset.
- The existing cells below separate the predictors (`x`) from the target (`y`), split the data into training and testing sets, and standardize the predictors. The scaler is fitted on the training data only, then applied to the test data to avoid data leakage.

## R square and adjusted R square

Formula

**R^2 = 1 - SSR/SST**

R^2	=	coefficient of determination
SSR	=	sum of squares of residuals
SST	=	total sum of squares

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print(score)
```

**Adjusted R2 = 1 – [(1-R2)*(n-1)/(n-k-1)]**

where:

R2: The R2 of the model
n: The number of observations
k: The number of predictor variables

```python
#display adjusted R-squared
1 - (1-score)*(len(y_test)-1)/(len(y_test)-x_test.shape[1]-1)
```
